```python
# EJEMPLO N° 4: VIAJES ANUALES DE SUBTE DESDE AÑO 201

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter
```

```python
df = pd.read_excel ("viajes_anual.xlsx")
df
```

```python
# Renombro solamente un campo.

df_rename = {"year": "año"}

df.rename (columns = df_rename, inplace = True)
df
```

```python
df.info ()
```

```python
df_filtro1 = df.groupby ("linea", as_index = False)["total"].sum ()

sns.set_style ("darkgrid")
sns.barplot (data = df_filtro1, x = "linea", y = "total", hue = "linea", palette = "rocket")
plt.title ("Cantidad total de pasajeros por línea durante los años 2013 a 2020", fontsize = 16)
plt.xlabel ("Línea")
plt.ylabel ("Pasajeros")
plt.ticklabel_format(style='plain', axis='y') 
plt.show ()
```

Se verifica que, durante el período 2013-2020, fue la línea B la que mayor cantidad de pasajeros recibió, seguida de la línea D. Por su parte, las estaciones E y H fueron las que menor cantidad de pasajeros recibieron durante ese mismo período.

```python
sns.set_style ("darkgrid")
sns.lineplot (data = df, x = "año", y ="total", hue = "linea", linewidth = 2, palette = "rocket")
plt.title ("Evolución de la cantidad total de pasajeros por línea (2013-2020)", fontsize = 16)
plt.xlabel ("Año")
plt.ylabel ("Pasajeros")
plt.ticklabel_format (style = "plain", axis = "y")
plt.show ()
```

```python
df_filtro2 = df[(df["linea"] == "H")]

sns.set_style ("darkgrid")
sns.lineplot (data = df_filtro2, x = "año", y = "total", color = "#ad1457", linewidth = 2)
plt.title ("Evolución de la cntidad de pasajeros en la línea H (2013-2020)", fontsize = 16)
plt.xlabel ("Años")
plt.ylabel ("Pasajeros")
plt.ticklabel_format(style='plain', axis='y') 
plt.show ()
```

Se verifica que desde el año 2015 a 2018 la cantidad de pasajeros se mantuvo siempre aumentado, decreciendo sólo entre el período 2018-2019, hasta alcanzar en este último año la mayor cantidad de pasajeros. Luego, se observa un fuerte decrecimiento del número de pasajeros, probablemente debido a la pandemia del COVID-19.

```python
df_filtro3 = df[(df["año"] == 2014)].sort_values ("total", ascending = False)

sns.set_style ("darkgrid")
sns.barplot (data = df_filtro3, x = "linea", y = "total", hue = "linea", palette = "rocket")
plt.title ("Cantidad total de pasajeros por línea en el año 2014", fontsize = 16)
plt.xlabel ("Línea")
plt.ylabel ("Pasajeros")
plt.ticklabel_format(style='plain', axis='y') 
plt.show ()
```

Se verifica que la línea B es la línea de pasajeros que más cantidad de pasajeros recibió en el año 2013, seguida por la línea D con valores casi similares. Por su parte, la línea H fue la que menor cantidad de pasajeros recibió.

```python
sns.set_style ("darkgrid")
sns.boxenplot (data = df, x = "linea", y = "total", hue = "linea", palette = "rocket")
plt.title ("")
plt.xlabel ("Línea")
plt.ylabel ("Pasajeros")
plt.ticklabel_format (style = "plain", axis = "y")
plt.show ()
```

Todas las líneas presentan valores atípicos. La línea E y H presentan una mediana menor al resto de las líneas.
